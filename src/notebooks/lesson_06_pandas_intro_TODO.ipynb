## Краткая теория по ноутбуку

На этом занятии мы переходим от простых структур Python к работе с настоящими табличными данными.

Главный инструмент занятия — библиотека **pandas**.  
Она позволяет:
- загружать таблицы из CSV и Excel
- хранить данные в структуре **DataFrame**
- быстро просматривать строки и столбцы
- выбирать нужные столбцы для анализа

### Что важно понять

- **CSV** — это текстовый файл, где значения разделены запятыми.
- **Excel-файл** (`.xlsx`) — это табличный файл с листами, форматированием и формулами.
- **`pandas`** — библиотека для анализа данных в Python.
- **`DataFrame`** — таблица в Python, похожая на лист Excel.
- **Строки** — наблюдения или записи.
- **Столбцы** — признаки, поля, характеристики.

### Основные команды занятия

- `import pandas as pd` — подключить библиотеку
- `pd.read_csv("file.csv")` — загрузить CSV
- `pd.read_excel("file.xlsx", sheet_name="...")` — загрузить Excel-лист
- `df.head()` — посмотреть первые строки
- `df.tail()` — посмотреть последние строки
- `df.shape` — узнать размер таблицы
- `df.columns` — получить названия столбцов
- `df["column"]` — выбрать один столбец
- `df[["col1", "col2"]]` — выбрать несколько столбцов

## Практическая ячейка 1. Импорт `pandas` и загрузка CSV

### Что делаем
Создаём небольшой CSV-файл прямо в ноутбуке и считываем его в `DataFrame`.

### Функции, методы и синтаксис
- `import pandas as pd` — подключение библиотеки
- `from pathlib import Path` — удобная работа с файлами
- `Path(...).write_text(...)` — сохранить текст в файл
- `pd.read_csv("имя_файла.csv")` — загрузить CSV в `DataFrame`

### На что обратить внимание
CSV — это обычный текстовый файл.  
После загрузки `pandas` превращает его в таблицу, с которой можно работать как с Excel-данными.


```python
import pandas as pd
from pathlib import Path

csv_text = """product_id,product_name,category,calories,protein,fat,carbs,fiber
1,Яблоко,Фрукты,52,0.3,0.2,14,2.4
2,Куриная грудка,Мясо,165,31,3.6,0,0
3,Овсянка,Крупы,389,16.9,6.9,66.3,10.6
4,Творог 5%,Молочные,121,17,5,3,0
5,Лосось,Рыба,208,20,13,0,0
6,Брокколи,Овощи,34,2.8,0.4,7,2.6
7,Миндаль,Орехи,579,21,49,22,12
8,Банан,Фрукты,89,1.1,0.3,23,2.6"""

Path("../data/lesson_06_nutrition_data.csv").write_text(csv_text, encoding="utf-8")

df_csv = pd.read_csv("../data/lesson_06_nutrition_data.csv")
df_csv
```
