# Занятие 8. Фильтрация, сортировка и вычисляемые столбцы

## Краткая теория по ноутбуку

На этом занятии мы продолжаем анализировать табличные данные в `pandas`.  
Наша цель — научиться:
- **фильтровать** строки по одному и нескольким условиям;
- **сортировать** таблицу по нужному столбцу;
- **создавать новые столбцы** для вычислений;
- делать короткий мини-анализ по заказам.

## Что важно понять
- **Фильтрация** — это отбор только тех строк, которые соответствуют условию.
- **Сортировка** помогает увидеть самые большие, самые дорогие или самые важные значения.
- **Вычисляемый столбец** — это новый столбец, который мы создаём на основе уже существующих данных.
- В Python это делается быстро и повторяемо, а в Excel — через фильтры, сортировку и формулы.

## Сегодня будем использовать
- `pd.read_excel(...)`
- `df[...]`
- `==`
- `&`
- `sort_values(...)`
- `df["new_column"] = ...`
- `sum()`


## Практическая ячейка 1. Загружаем Excel и смотрим на структуру таблицы

### Что делаем
Открываем Excel-файл занятия, считываем лист `sales_data` и смотрим:
- размер таблицы;
- названия столбцов;
- первые строки.

### Функции, методы и синтаксис
- `import pandas as pd` — подключаем библиотеку `pandas`.
- `Path("имя_файла.xlsx")` — создаём удобный путь к файлу.
- `pd.read_excel(file_path, sheet_name="sales_data", header=1)` — читаем лист Excel в `DataFrame`.
- `df.shape` — показывает количество строк и столбцов.
- `df.columns` — показывает названия столбцов.
- `df.head()` — показывает первые строки таблицы.


```python
import pandas as pd
from pathlib import Path

file_path = Path("../../xls/lesson_08_filter_sort_calc.xlsx")

df = pd.read_excel(file_path, sheet_name="sales_data", header=1)

print("Размер таблицы:", df.shape)
print("\nНазвания столбцов:")
print(list(df.columns))

df.head()
```

## Практическая ячейка 2. Фильтрация по одному условию

### Что делаем
Оставляем только те строки, где категория товара — `Ноутбуки`.

### Функции, методы и синтаксис
- `df["column"]` — обращение к столбцу таблицы.
- `df["column"] == "значение"` — создаёт булеву маску: `True` там, где условие выполняется.
- `df[маска]` — возвращает только строки, которые подходят под условие.
- `len(df_filtered)` — показывает количество строк после фильтрации.


```python
notebooks_df = df[df["category"] == "Ноутбуки"]

print("Количество заказов в категории 'Ноутбуки':", len(notebooks_df))
notebooks_df
```

## Практическая ячейка 3. Фильтрация по нескольким условиям

### Что делаем
Оставляем только те заказы, которые одновременно:
- сделаны в городе `Москва`;
- пришли через канал `Онлайн`.

### Функции, методы и синтаксис
- `(условие1) & (условие2)` — логическое И для двух условий в `pandas`.
- Каждое условие нужно брать в круглые скобки.
- `df[(... ) & (... )]` — фильтрация по нескольким условиям сразу.


```python
moscow_online_df = df[(df["city"] == "Москва") & (df["channel"] == "Онлайн")]

print("Количество заказов из Москвы через онлайн-канал:", len(moscow_online_df))
moscow_online_df
```

## Практическая ячейка 4. Сортировка таблицы

### Что делаем
Сортируем заказы по столбцу `price` по убыванию, чтобы увидеть самые дорогие позиции.

### Функции, методы и синтаксис
- `df.sort_values(by="column")` — сортирует таблицу по выбранному столбцу.
- `ascending=False` — сортировка по убыванию.
- `head()` — позволяет быстро посмотреть первые строки результата.


```python
sorted_by_price = df.sort_values(by="price_per_unit", ascending=False)

print("Первые 5 заказов после сортировки по цене:")
sorted_by_price.head()
```

## Практическая ячейка 5. Создаём новый столбец `revenue` и решаем мини-задачу

### Что делаем
Создаём новый столбец `revenue` по формуле:

`revenue = quantity * price`

После этого:
- считаем общую выручку;
- оставляем только крупные заказы, где `revenue >= 100000`;
- сортируем их по `revenue` по убыванию.

### Функции, методы и синтаксис
- `df.copy()` — создаёт копию таблицы.
- `df["new_column"] = ...` — создаёт новый столбец.
- `df["quantity"] * df["price"]` — поэлементное умножение столбцов.
- `df["revenue"].sum()` — считает сумму по новому столбцу.
- `df[df["revenue"] >= 100000]` — фильтрация по вычисленному столбцу.
- `sort_values(by="revenue", ascending=False)` — сортировка по выручке.


```python
df_calc = df.copy()

df_calc["revenue"] = df_calc["quantity"] * df_calc["price_per_unit"]

big_orders = df_calc[df_calc["revenue"] >= 100000].sort_values(
    by="revenue", ascending=False
)

print("Общая выручка:", df_calc["revenue"].sum())
print("Количество крупных заказов:", len(big_orders))

big_orders
```

## Ячейка 6. Тест и самопроверка

### Что делаем
Проверяем, что вы умеете:
- загружать Excel-лист;
- фильтровать таблицу по одному условию;
- фильтровать по нескольким условиям;
- сортировать данные;
- создавать вычисляемый столбец;
- считать выручку и находить крупные заказы.

### Функции, методы и синтаксис
- `assert условие` — если условие ложное, Python покажет ошибку.
- `shape`, `sort_values()`, `sum()` — основные проверки по теме занятия.


```python
import pandas as pd

test_df = pd.read_excel("../../xls/lesson_08_filter_sort_calc.xlsx", sheet_name="sales_data", header=1)
print(test_df.shape)
assert test_df.shape == (5000, 15)
assert list(test_df.columns) == [
    "order_id",
    "date",
    "day",
    "month",
    "manager",
    "city",
    "region",
    "category",
    "product",
    "client_type",
    "quantity",
    "price_per_unit",
    "revenue",
    "channel",
    "status",
]

test_notebooks = test_df[test_df["category"] == "Ноутбуки"]
print(len(test_notebooks))
assert len(test_notebooks) == 813

test_moscow_online = test_df[
    (test_df["city"] == "Москва") & (test_df["channel"] == "Интернет-магазин")
]
print(len(test_moscow_online))
assert len(test_moscow_online) == 156

test_sorted = test_df.sort_values(by="price_per_unit", ascending=False)
print(test_sorted[["order_id", "price_per_unit"]])
assert int(test_sorted.iloc[0]["order_id"]) == 180
assert int(test_sorted.iloc[0]["price_per_unit"]) == 227200

test_calc = test_df.copy()
test_calc["revenue"] = test_calc["quantity"] * test_calc["price_per_unit"]
print(int(test_calc["revenue"].sum()))
assert int(test_calc["revenue"].sum()) == 2612584200

test_big_orders = test_calc[test_calc["revenue"] >= 8300000].sort_values(by="revenue", ascending=False)
print(len(test_big_orders))
print(list(test_big_orders["order_id"]))
assert len(test_big_orders) == 3
assert list(test_big_orders["order_id"]) == [2176, 1711, 2930]

print("Тест пройден: фильтрация, сортировка и вычисления выполнены корректно.")
```
