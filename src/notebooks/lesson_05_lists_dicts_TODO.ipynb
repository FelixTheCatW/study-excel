# Занятие 5. Списки, словари и табличное мышление

## Краткая теория по ноутбуку

На этом занятии мы переходим от отдельных переменных к структурам данных.

Если переменная хранит одно значение, то более сложные структуры позволяют хранить:
- несколько элементов сразу
- запись с набором полей
- маленькую таблицу прямо в Python

### Что важно понять
В Python часто используются:

- **Список (`list`)** — упорядоченный набор элементов
- **Словарь (`dict`)** — набор пар `ключ: значение`
- **Список словарей** — удобная модель мини-таблицы, где каждый словарь похож на одну строку

### Базовые термины
- **Элемент списка** — одно значение внутри списка
- **Индекс** — номер позиции в списке
- **Ключ** — имя поля в словаре
- **Значение** — данные, связанные с ключом
- **Запись** — один объект со своими полями
- **Структура данных** — способ хранения информации в удобной форме

5. Мини-задачу по заказам и выручке

## Практическая ячейка 4. Список словарей как мини-таблица

### Что делаем

### Функции, методы и синтаксис
- `[{...}, {...}]` — список словарей

## Практическая ячейка 5. Мини-задача: заказ и общая выручка

Используем список словарей, чтобы посчитать общую выручку по всем заказам.

### Функции, методы и синтаксис
- список словарей
- цикл `for`
- накопление суммы
- доступ к полям записи по ключам

### На что обратить внимание
Это уже почти ручная версия будущей таблицы `DataFrame`.

```python
products = [
    {
        "product_name": "Банан",
        "category": "Фрукты",
        "calories": 89,
        "protein": 1.1,
        "weight": 320,
    },
    {
        "product_name": "Миндаль",
        "category": "Орехи",
        "calories": 579,
        "protein": 21,
        "weight": 60,
    },
    {
        "product_name": "Брокколи",
        "category": "Овощи",
        "calories": 34,
        "protein": 2.8,
        "weight": 280,
    },
]

total_calories = 0
for row in products:
    calories = row["calories"] * row["weight"] / 100
    total_calories += calories
    # print("Блюдо:", row["product_name"], "| общая калорийность:", total)
    
print("Количество блюд:", len(products))
print("Общий калораж:", round(total_calories))
```
