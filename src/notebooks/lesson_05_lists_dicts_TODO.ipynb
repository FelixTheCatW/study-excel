# Занятие 5. Списки, словари и табличное мышление

## Краткая теория по ноутбуку

На этом занятии мы переходим от отдельных переменных к структурам данных.

Если переменная хранит одно значение, то более сложные структуры позволяют хранить:
- несколько элементов сразу
- запись с набором полей
- маленькую таблицу прямо в Python

### Что важно понять
В Python часто используются:

- **Список (`list`)** — упорядоченный набор элементов
- **Словарь (`dict`)** — набор пар `ключ: значение`
- **Список словарей** — удобная модель мини-таблицы, где каждый словарь похож на одну строку

### Базовые термины
- **Элемент списка** — одно значение внутри списка
- **Индекс** — номер позиции в списке
- **Ключ** — имя поля в словаре
- **Значение** — данные, связанные с ключом
- **Запись** — один объект со своими полями
- **Структура данных** — способ хранения информации в удобной форме
