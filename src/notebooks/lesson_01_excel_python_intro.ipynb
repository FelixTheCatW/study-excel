## Практическая ячейка 1. Переменные и типы данных

### Что делаем
Создаём несколько переменных и смотрим, какие типы данных они содержат.

### Функции, методы и синтаксис
- `имя_переменной = значение` — создаёт переменную
- `print(...)` — выводит значение на экран
- `type(...)` — показывает тип данных объекта
- `type(x).__name__` — позволяет вывести короткое имя типа: `int`, `float`, `str`, `bool`

### На что обратить внимание
В аналитике очень важно понимать, с чем вы работаете: с числом, текстом или логическим значением.

```python
client_name = "ООО Альфа"
orders_count = 12
avg_check = 1850.5
is_vip = True

print("Название клиента:", client_name, "| тип:", type(client_name).__name__)
print("Количество заказов:", orders_count, "| тип:", type(orders_count).__name__)
print("Средний чек:", avg_check, "| тип:", type(avg_check).__name__)
print("VIP-клиент:", is_vip, "| тип:", type(is_vip).__name__)
```

## Практическая ячейка 2. Арифметические операции

### Что делаем
Рассчитываем стоимость заказа, сумму скидки и итоговую сумму.

### Функции, методы и синтаксис
- `+` — сложение
- `-` — вычитание
- `*` — умножение
- `/` — деление
- Переменные можно использовать внутри формул, чтобы строить расчёты по шагам

### На что обратить внимание
Лучше делать расчёты поэтапно: сначала промежуточная сумма, потом скидка, потом итог. Так код легче читать и проверять.

```python
price = 2500
quantity = 4
discount_rate = 0.10

subtotal = price * quantity
discount_amount = subtotal * discount_rate
total = subtotal - discount_amount

print("Цена:", price)
print("Количество:", quantity)
print("Сумма без скидки:", subtotal)
print("Сумма скидки:", discount_amount)
print("Итог к оплате:", total)
```

## Практическая ячейка 3. Преобразование типов

### Что делаем
Берём числа, записанные как текст, и преобразуем их в числовой формат для расчётов.

### Функции, методы и синтаксис
- `int("10")` — превращает текст в целое число
- `float("19.95")` — превращает текст в число с дробной частью
- `str(25)` — превращает число в текст

### На что обратить внимание
В реальных данных из Excel, CSV, форм или CRM числа часто приходят как строки. Пока тип не преобразован, корректные вычисления невозможны.

```python
price_text = "1999.90"
qty_text = "3"

price_num = float(price_text)
qty_num = int(qty_text)

total_sum = price_num * qty_num

print("Исходная цена как текст:", price_text, "| тип:", type(price_text).__name__)
print("Цена после преобразования:", price_num, "| тип:", type(price_num).__name__)
print("Количество после преобразования:", qty_num, "| тип:", type(qty_num).__name__)
print("Итоговая сумма:", total_sum)
```

## Практическая ячейка 4. Табличное мышление: список значений

### Что делаем
Храним данные о товарах в списках и выводим их как простую таблицу.

### Функции, методы и синтаксис
- `[]` — создаёт список
- `len(список)` — количество элементов
- `range(...)` — диапазон индексов
- `for ... in ...` — цикл перебора
- `список[i]` — обращение к элементу по индексу

### На что обратить внимание
Список в Python — это первый шаг к табличной логике. Позже мы перейдём к `pandas`, где таблицы будут уже полноценными.

```python
products = ["Ноутбук", "Мышь", "Клавиатура"]
prices = [250000, 4500, 12000]
quantities = [2, 5, 3]

print("Мини-таблица товаров:")
for i in range(len(products)):
    row_total = prices[i] * quantities[i]
    print(f"{i + 1}. {products[i]} | цена: {prices[i]} | количество: {quantities[i]} | сумма: {row_total}")

print("\nКоличество товаров:", len(products))
print("Максимальная цена:", max(prices))
```

## Практическая ячейка 5. Мини-задача: расчёт выручки по заказу

### Что делаем
Создаём мини-модель одного заказа и считаем:
- выручку без скидки
- сумму скидки
- итоговую сумму

### Функции, методы и синтаксис
- Словарь создаётся через `{ключ: значение}`
- Доступ к значению словаря: `словарь["ключ"]`
- Формулы строятся из уже созданных значений

### На что обратить внимание
Такой подход — основа будущих бизнес-расчётов, аналитики продаж и подготовки данных для ИИ-систем.

```python
order = {
    "product": "Монитор",
    "price": 65000,
    "quantity": 2,
    "discount_rate": 0.07
}

revenue = order["price"] * order["quantity"]
discount = revenue * order["discount_rate"]
final_total = revenue - discount

print("Товар:", order["product"])
print("Выручка без скидки:", revenue)
print("Сумма скидки:", discount)
print("Итог к оплате:", final_total)
```

## Ячейка 6. Тест и самопроверка

### Что делаем
Проверяем, что вы понимаете:
- типы данных
- арифметику
- преобразование типов
- работу со списками
- расчёт итоговой суммы

### Функции, методы и синтаксис
- `assert условие` — проверка
- если условие истинно, программа идёт дальше
- если условие ложно, Python выдаёт ошибку, и это помогает быстро найти проблему

```python
# 1. Проверка типов данных
name = "Аналитик"
age = 30
salary = 125000.0
is_remote = False

assert type(name).__name__ == "str"
assert type(age).__name__ == "int"
assert type(salary).__name__ == "float"
assert type(is_remote).__name__ == "bool"

# 2. Проверка арифметики
price = 3000
quantity = 5
subtotal = price * quantity
assert subtotal == 15000

# 3. Проверка преобразования типов
price_text = "2500.5"
quantity_text = "2"
result = float(price_text) * int(quantity_text)
assert result == 5001.0

# 4. Проверка списка
items = ["A", "B", "C", "D"]
assert len(items) == 4
assert items[0] == "A"
assert items[-1] == "D"

# 5. Проверка мини-задачи
order = {"price": 10000, "quantity": 3, "discount_rate": 0.10}
revenue = order["price"] * order["quantity"]
discount = revenue * order["discount_rate"]
final_total = revenue - discount

assert revenue == 30000
assert discount == 3000
assert final_total == 27000

print("Тест пройден успешно. Все проверки выполнены.")
```
