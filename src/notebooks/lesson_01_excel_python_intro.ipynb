## Практическая ячейка 1. Переменные и типы данных

### Что делаем
Создаём несколько переменных и смотрим, какие типы данных они содержат.

### Функции, методы и синтаксис
- `имя_переменной = значение` — создаёт переменную
- `print(...)` — выводит значение на экран
- `type(...)` — показывает тип данных объекта
- `type(x).__name__` — позволяет вывести короткое имя типа: `int`, `float`, `str`, `bool`

### На что обратить внимание
В аналитике очень важно понимать, с чем вы работаете: с числом, текстом или логическим значением.

```python
product_name = "Говядина (постная)"
servings_count = 12
daily_intake = 2400
is_gain = False

print("Название продукта:", product_name, "| тип:", type(product_name).__name__)
print("Количество порций:", servings_count, "| тип:", type(servings_count).__name__)
print("Суточный калораж:", daily_intake, "| тип:", type(daily_intake).__name__)
print("Набор массы:", is_gain, "| тип:", type(is_gain).__name__)
```

## Практическая ячейка 2. Арифметические операции

### Что делаем
Рассчитываем стоимость заказа, сумму скидки и итоговую сумму.

### Функции, методы и синтаксис
- `+` — сложение
- `-` — вычитание
- `*` — умножение
- `/` — деление
- Переменные можно использовать внутри формул, чтобы строить расчёты по шагам

### На что обратить внимание
Лучше делать расчёты поэтапно: сначала промежуточная сумма, потом скидка, потом итог. Так код легче читать и проверять.

```python
calories_per_100g = 165
weight_grams = 150
loss_factor = 0.10

subtotal = (calories_per_100g / 100) * weight_grams

loss_amount = subtotal * loss_factor

total = subtotal - loss_amount

print("Калорийность на 100 г:", calories_per_100g)
print("Вес порции (г):", weight_grams)
print("Общие калории в порции:", subtotal)
print("Потери при готовке (ккал):", loss_amount)
print("Итоговые усвоенные калории:", total)
```

## Практическая ячейка 3. Преобразование типов

### Что делаем
Берём числа, записанные как текст, и преобразуем их в числовой формат для расчётов.

### Функции, методы и синтаксис
- `int("10")` — превращает текст в целое число
- `float("19.95")` — превращает текст в число с дробной частью
- `str(25)` — превращает число в текст

### На что обратить внимание
В реальных данных из Excel, CSV, форм или CRM числа часто приходят как строки. Пока тип не преобразован, корректные вычисления невозможны.

```python
calories_text = "165.5"
weight_text = "200"

calories_num = float(calories_text)
weight_num = int(weight_text)

total_calories = (calories_num / 100) * weight_num

print(
    "Исходная калорийность как текст:",
    calories_text,
    "| тип:",
    type(calories_text).__name__,
)
print(
    "Калорийность после преобразования:",
    calories_num,
    "| тип:",
    type(calories_num).__name__,
)
print("Вес после преобразования:", weight_num, "| тип:", type(weight_num).__name__)
print("Общая калорийность порции:", total_calories)
```

## Практическая ячейка 4. Табличное мышление: список значений

### Что делаем
Храним данные о товарах в списках и выводим их как простую таблицу.

### Функции, методы и синтаксис
- `[]` — создаёт список
- `len(список)` — количество элементов
- `range(...)` — диапазон индексов
- `for ... in ...` — цикл перебора
- `список[i]` — обращение к элементу по индексу

### На что обратить внимание
Список в Python — это первый шаг к табличной логике. Позже мы перейдём к `pandas`, где таблицы будут уже полноценными.

```python
product_names = ["Гречка", "Рис", "Овсянка"]
calories_per_100 = [343, 340, 360]
portion_weights = [150, 200, 180]

print("Таблица продуктов:")
for i in range(len(product_names)):
    portion_cal = (calories_per_100[i] / 100) * portion_weights[i]
    print(
        f"{i + 1}. {product_names[i]} | калорийность: {calories_per_100[i]} ккал/100г | вес порции: {portion_weights[i]} г | калорий порции: {portion_cal:.1f} ккал"
    )

print("\nКоличество продуктов:", len(product_names))
print("Максимальная калорийность на 100 г:", max(calories_per_100))
```

## Практическая ячейка 5. Мини-задача: расчёт выручки по заказу

### Что делаем
Создаём мини-модель одного заказа и считаем:
- выручку без скидки
- сумму скидки
- итоговую сумму

### Функции, методы и синтаксис
- Словарь создаётся через `{ключ: значение}`
- Доступ к значению словаря: `словарь["ключ"]`
- Формулы строятся из уже созданных значений

### На что обратить внимание
Такой подход — основа будущих бизнес-расчётов, аналитики продаж и подготовки данных для ИИ-систем.

```python
product = {
    "name": "Лосось",
    "calories_per_100g": 208,
    "weight_grams": 120,
    "loss_factor": 0.08,
}

# Расчёты:
total_raw_cal = (product["calories_per_100g"] / 100) * product["weight_grams"]
loss_cal = total_raw_cal * product["loss_factor"]
final_cal = total_raw_cal - loss_cal

print("Продукт:", product["name"])
print("Общие калории (сырые):", round(total_raw_cal))
print("Потери при готовке (ккал):", round(loss_cal))
print("Итоговые калории (усвоенные):", round(final_cal))
```

## Ячейка 6. Тест и самопроверка

### Что делаем
Проверяем, что вы понимаете:
- типы данных
- арифметику
- преобразование типов
- работу со списками
- расчёт итоговой суммы

### Функции, методы и синтаксис
- `assert условие` — проверка
- если условие истинно, программа идёт дальше
- если условие ложно, Python выдаёт ошибку, и это помогает быстро найти проблему

```python
# 1. Проверка типов данных
product = "Кефир"
fat_percent = 2.5
servings = 2
is_low_fat = True

assert type(product).__name__ == "str"
assert type(fat_percent).__name__ == "float"
assert type(servings).__name__ == "int"
assert type(is_low_fat).__name__ == "bool"

# 2. Проверка арифметики (общая калорийность)
cal_per_100 = 56
weight = 200
total_cal = (cal_per_100 / 100) * weight
assert total_cal - 112.0 < 1e-6

# 3. Проверка преобразования типов
cal_text = "210.7"
weight_text = "150"
result = float(cal_text) / 100 * int(weight_text)
assert result - 316.05 < 1e-6

# 4. Проверка списка
foods = ["Яблоко", "Банан", "Апельсин"]
assert len(foods) == 3
assert foods[0] == "Яблоко"
assert foods[-1] == "Апельсин"

# 5. Проверка мини-задачи (словарь)
meal = {"calories_per_100g": 350, "weight_grams": 80, "loss_factor": 0.05}
raw = (meal["calories_per_100g"] / 100) * meal["weight_grams"]
lost = raw * meal["loss_factor"]
final = raw - lost

assert raw == 280.0
assert lost == 14.0
assert final == 266.0

print("Тест пройден успешно. Все проверки выполнены.")
```
