- **Срез** — часть строки

## Практическая ячейка 1. Строки и длина текста

### Что делаем
Создаём текстовые переменные и измеряем длину строки.

### Функции, методы и синтаксис
- `"текст"` — строковое значение
- `len(строка)` — длина строки
- `print(...)` — вывод результата

### На что обратить внимание
Лишние пробелы тоже считаются символами. Это важно при проверке качества данных.

```python
client_name = "  Ирина Смирнова  "
product_code = "AI-COURSE-2025"
city = "Москва"

print("Исходное имя клиента:", repr(client_name))
print("Длина строки client_name:", len(client_name))
print("Код продукта:", product_code)
print("Длина кода продукта:", len(product_code))
print("Город:", city)
```

## Практическая ячейка 2. Индексация и срезы

### Что делаем
Получаем отдельные символы и части строки.

### Функции, методы и синтаксис
- `строка[0]` — первый символ
- `строка[-1]` — последний символ
- `строка[0:3]` — срез от первого до третьего символа
- `строка[3:]` — срез от позиции 3 до конца

### На что обратить внимание
Индексация в Python начинается с нуля.

```python
product_code = "AI-COURSE-2025"

first_symbol = product_code[0]
last_symbol = product_code[-1]
prefix = product_code[0:2]
course_word = product_code[3:9]
year_part = product_code[-4:]

print("Первый символ:", first_symbol)
print("Последний символ:", last_symbol)
print("Префикс:", prefix)
print("Часть с названием:", course_word)
print("Год:", year_part)
```

## Практическая ячейка 3. Очистка текста и методы строк

### Что делаем
Очищаем строку от лишних пробелов, меняем регистр и заменяем символы.

### Функции, методы и синтаксис
- `.strip()` — удаляет пробелы в начале и в конце строки
- `.lower()` — переводит текст в нижний регистр
- `.upper()` — переводит текст в верхний регистр
- `.replace("старое", "новое")` — заменяет фрагмент строки

### На что обратить внимание
Это один из самых частых шагов в подготовке данных.

```python
raw_category = "  Аналитика-ДАННЫХ  "

clean_category = raw_category.strip()
lower_category = clean_category.lower()
upper_category = clean_category.upper()
replaced_category = lower_category.replace("-", " ")

print("Исходная строка:", repr(raw_category))
print("После strip():", repr(clean_category))
print("После lower():", lower_category)
print("После upper():", upper_category)
print("После replace():", replaced_category)
```

## Практическая ячейка 4. Склейка строк и форматирование текста

### Что делаем
Объединяем несколько текстовых частей в одно осмысленное значение.

### Функции, методы и синтаксис
- `+` — склейка строк
- `f"..."` — форматированная строка
- можно собирать ярлыки, подписи, e-mail и человекочитаемые описания

### На что обратить внимание
Склейка часто нужна для создания отчётных меток, идентификаторов и подписей.

```python
first_name = "Ирина"
last_name = "Смирнова"
department = "AI Lab"

full_name = first_name + " " + last_name
email = first_name.lower() + "." + last_name.lower() + "@example.com"
label = f"{full_name} | отдел: {department}"

print("Полное имя:", full_name)
print("E-mail:", email)
print("Подпись:", label)
```

## Практическая ячейка 5. Мини-задача: очистка карточки товара

### Что делаем
Берём «грязные» текстовые данные и превращаем их в аккуратные значения для анализа.

### Функции, методы и синтаксис
- `.strip()` для удаления лишних пробелов
- `.lower()` и `.upper()` для нормализации регистра
- `.replace()` для замены символов
- `f"..."` для итогового ярлыка

### На что обратить внимание
Это уже очень похоже на реальную задачу подготовки данных.

```python
raw_product_name = "  ноутбук-Игровой  "
raw_brand = "  tecHVision "
raw_category = "  ЭЛЕКТРОНИКА "

product_name = raw_product_name.strip().replace("-", " ")
brand = raw_brand.strip().upper()
category = raw_category.strip().lower()
card_label = f"{brand} | {product_name} | категория: {category}"

print("Очищенное название товара:", product_name)
print("Бренд:", brand)
print("Категория:", category)
print("Итоговая карточка:", card_label)
```

## Ячейка 6. Тест и самопроверка

### Что делаем
Проверяем:
- длину строки
- индексацию и срезы
- очистку пробелов
- смену регистра
- замену символов
- склейку строк

### Функции, методы и синтаксис
- `assert условие` — быстрая проверка правильности результата

```python
# 1. Длина строки
text = "AI system"
assert len(text) == 9

# 2. Индексация и срезы
code_value = "ML-2025"
assert code_value[0] == "M"
assert code_value[-4:] == "2025"

# 3. Очистка пробелов
raw_name = "  Ivan  "
assert raw_name.strip() == "Ivan"

# 4. Регистр
category = "DataScience"
assert category.lower() == "datascience"
assert category.upper() == "DATASCIENCE"

# 5. Замена символов
tag = "ai-course-basic"
assert tag.replace("-", "_") == "ai_course_basic"

# 6. Склейка строк
first = "Data"
second = "Analyst"
full = first + " " + second
assert full == "Data Analyst"

print("Тест пройден успешно. Все проверки выполнены.")
```
